# 클래스의 정의

```python
#클래스 정의
class Student :
    def __init__(self, name, age):
        self.name = name + '@'
        self.age = age
    def introduce(self):
        print(f'안녕하세요. 제 이름은 {self.name} 입니다. {self.age}살입니다.')
```

# 객체 생성 & 호출

```python
s1 = Student("alice", 25)
s1.introduce()

s2 = Student("billy", 35)
s2.introduce()
```

```python
#클래스 정의
class Calculator:
    def __init__(self, number_list):
        self.number_list = number_list
    def sum(self):
        return sum(self.number_list)
    def avg(self):
        return sum(self.number_list)/len(self.number_list)
```

```python
#객체 생성
cal1 = Calculator([1,2,3,4,5])
print(cal1.sum())
print(cal1.avg())

cal2 = Calculator([6,7,8,9,0])
print(cal2.sum(), cal1.avg())
```

다형성
같은 메서드나 연산이 객체에 따라 다르게 동작할 수 있게 하는 특성

```python
class Animal:
    def make_sound(self):
        pass
class Dog(Animal):
    def make_sound(self):
        return "멍멍!"
class Cat(Animal):
    def make_sound(self):
        return "야옹!"
```

```python
animal = [Dog(), Cat(), Dog(), Cat()]

for ani in animal:
    print(type (ani))
    print(ani.make.sound())
```
