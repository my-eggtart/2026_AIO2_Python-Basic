```python
sentence = input("문장을 입력하세요: ")
words = ["파이썬", "코딩", "데이터", "ai"]

found_word = None

for word in words:
    if word in sentence:
        found_word = word
        break

if found_word:
    print(f"입력하신 문장에서 '{found_word}'를 찾았습니다.")
else:
    print("입력하신 문장에서 단어를 찾지 못했습니다.")
```
